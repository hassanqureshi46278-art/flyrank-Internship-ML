## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: which existing pages get reworked (content update, refreshed metadata, added sections) in the next content sprint, out of a much larger pool of candidate pages.

Who acts on it: a content/SEO strategist or account manager with limited hours each week, who needs a short, defensible shortlist rather than a spreadsheet of every page on the site.

Cost of a wrong call: there are two failure directions. A false positive — flagging a page that's actually fine — wastes a writer's hours and delays pages that genuinely needed the work. A false negative — skipping a page that's quietly losing search visibility or leaving demand on the table — means the client keeps losing traffic for another cycle before anyone notices. Because writer time is the scarce resource, precision at the top of the queue (the handful of pages someone actually reviews) matters more than covering every page — which is why the pipeline reports Precision@50 rather than a blanket accuracy score.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import os

LOCAL_PATH = "data/raw/content_refresh_anonymized.csv"
REMOTE_PATH = ("https://raw.githubusercontent.com/flyrank-bih/"
               "flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv")

df = pd.read_csv(LOCAL_PATH) if os.path.exists(LOCAL_PATH) else pd.read_csv(REMOTE_PATH)
print(f"Rows: {len(df):,} | Columns: {df.shape[1]}")
print(df.columns.tolist())

if {"days_since_update", "search_volume"}.issubset(df.columns):
    stale_cutoff = df["days_since_update"].quantile(0.75)
    demand_median = df["search_volume"].median()
    stale_high_demand = df[(df["days_since_update"] >= stale_cutoff) & (df["search_volume"] > demand_median)]
    print(f"Stale (top 25% days_since_update) AND above-median search_volume: "
          f"{len(stale_high_demand):,} ({len(stale_high_demand)/len(df):.1%})")

if {"search_volume", "clicks_90d"}.issubset(df.columns):
    demand_median = df["search_volume"].median()
    low_click_cutoff = df["clicks_90d"].quantile(0.25)
    demand_no_clicks = df[(df["search_volume"] > demand_median) & (df["clicks_90d"] < low_click_cutoff)]
    print(f"Above-median search_volume but bottom-quartile clicks_90d: "
          f"{len(demand_no_clicks):,} ({len(demand_no_clicks)/len(df):.1%})")
```
